# ModalAI module reference

Part-number lookup for the ModalAI catalogue, supporting
[the VOXL 2 Mini notes](07_voxl2_mini.ipynb). Everything a build touches is an
`M00xx` board, and the numbers are the only reliable identifier — the marketing
names collide (three different boards are called simply "VOXL 2").

Compiled 2026-07-31 from the live
[PCB catalog](https://docs.modalai.com/pcb-catalog/) plus the datasheet pages
enumerated from the docs site itself. **85 distinct part numbers**, of which 73
are current and 54 have a dedicated datasheet page at
`docs.modalai.com/<part number>/`. Every link below returned HTTP 200 when
checked.

Two things worth knowing before using the catalog:

- **The catalog table is stale.** 19 boards have live datasheet pages but do not
  appear in it at all — including `M0135`, `M0149`, `M0161`, `M0166`, `M0169`,
  `M0178`, `M0186`, `M0187`, `M0189` and `M0195`, which is most of the current
  sensor and adapter line. Search the docs by part number instead of trusting
  the table.
- **Weights are inconsistent.** Some are bare PCB, some include cables or lids
  (`M0030` is listed as "8.24g / 14g with modem + lid"). Treat them as
  indicative.

## Interposers, flexes, adapters

The part of the catalogue that quietly decides the mechanical layout.

**Connector families**, since the part descriptions assume familiarity with
them:

- **DF40** — Hirose 60-pin, 0.4 mm pitch (`DF40C-60DP-0.4V(51)`). The camera
  ports on VOXL 2 and the Mini, J6 and J7. Each carries two camera "groups".
- **AXT** — Panasonic P4S board-to-board, 36-pin (`AXT336124` board side,
  `AXT436124` mating). The image-sensor connector on VOXL 1 and VOXL Flight,
  and still the format most sensor modules and flexes terminate in.

"DF40C to AXT" adapters therefore exist to bridge generations: they break one group of
signals (CSI lanes, timers, resets) out of a 60-pin DF40 port into a 36-pin AXT
connector. `M0084` is the Y-variant that fans one port out to two sensors.

## Reproducing this list

The catalog is a Jekyll table on the docs site and the per-board pages are
enumerable from its navigation, so this is scriptable:

```bash
curl -s https://docs.modalai.com/pcb-catalog/ -o cat.html
# table rows use unclosed <td>/<tr>, so split rather than match pairs
# nav links of the form docs.modalai.com/<part number>/ give the datasheet pages
```

The docs site source is also in the local checkout at `voxl/documentation/`
(`docs/60-expansion-boards/99-pcb-catalog.md`), but note it lags the live site —
which is how the 19 missing entries were found.
